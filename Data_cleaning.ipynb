### Selecting Required Columns

- Selected 12 useful columns for our analysis.
- Removed unnecessary columns from the analysis.
- Created a new DataFrame named `df_analysis`.
- Kept the original dataset unchanged.
- Original dataset: 35 columns.
- New dataset: 12 columns.

```python
required_columns = [
    'Age',
    'Attrition',
    'OverTime',
    'MonthlyIncome',
    'JobSatisfaction',
    'WorkLifeBalance',
    'Department',
    'JobRole',
    'DistanceFromHome',
    'YearsAtCompany',
    'BusinessTravel',
    'MaritalStatus'
]

df_analysis = df[required_columns].copy()

df_analysis.head()
```

```python
print("Original DataFrame Shape:", df.shape)
print("New DataFrame Shape:", df_analysis.shape)
```

```python
numerical_columns = [
    'Age',
    'MonthlyIncome',
    'DistanceFromHome',
    'YearsAtCompany'
]

plt.figure(figsize=(12, 6))

df_analysis[numerical_columns].boxplot()

plt.title('Outlier Detection Using Boxplot')
plt.ylabel('Values')
plt.xticks(rotation=45)
plt.grid(False)
plt.show()
```

### Outlier Detection

- Used boxplots to find outliers.
- MonthlyIncome has some high values.
- YearsAtCompany has a few high values.
- Age and DistanceFromHome have no major outliers.
- Kept all the values because they may be correct.
